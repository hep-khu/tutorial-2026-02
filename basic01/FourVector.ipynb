```python
import math
```

```python
m = 0.105658

pt1 = 15.6571
eta1 = 0.316655
phi1 = 2.56765

pt2, eta2, phi2 = 30.6535, 1.66002, 2.71109
pt3, eta3, phi3 = 8.51924, -0.55083, -1.2578
pt4, eta4, phi4 = 25.0167, 2.32682, -0.408024
```

```python
def pxpypze(pt, eta, phi, m):
  px = pt*math.cos(phi)
  py = pt*math.sin(phi)
  pz = pt*math.sinh(eta)

  e = math.sqrt(px**2 + py**2 + pz**2 + m**2)

  return px, py, pz, e
```

```python
pxyze1 = pxpypze(pt1, eta1, phi1, m)
pxyze2 = pxpypze(pt2, eta2, phi2, m)
pxyze3 = pxpypze(pt3, eta3, phi3, m)
pxyze4 = pxpypze(pt4, eta4, phi4, m)
```

```python
px = pxyze1[0] + pxyze2[0] + pxyze3[0] + pxyze4[0]
py = pxyze1[1] + pxyze2[1] + pxyze3[1] + pxyze4[1]
pz = pxyze1[2] + pxyze2[2] + pxyze3[2] + pxyze4[2]
e = pxyze1[3] + pxyze2[3] + pxyze3[3] + pxyze4[3]
```

```python
m0 = math.sqrt(e**2 - px**2 - py**2 - pz**2)
```
